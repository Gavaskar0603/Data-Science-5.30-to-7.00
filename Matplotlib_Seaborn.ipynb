```python
import pandas as pd
```

```python
d = pd.read_csv("/content/student_performance.csv")
d
```

```python
d.columns
```

```python
import matplotlib.pyplot as plt
```

```python
d.columns[d.isnull().any()]
```

```python
d.groupby("grade")[["total_score","weekly_self_study_hours"]].mean().plot(kind="bar")
plt.title("Bar chart")
plt.xlabel("Grade")
plt.ylabel("Score")
plt.show()
```

```python
dict_d = {"choclate":["Diarymilk","5Star","KitKat","Milkybar"],
          "Sale":[5000,2500,8000,1800]}
dict_d
```

```python
plt.bar(dict_d["choclate"],dict_d["Sale"],color=["violet","yellow","red","green"])
plt.show()
```

```python
plt.hist(d["total_score"])
plt.xlabel("Score")
plt.ylabel("count")
plt.show()
```

```python
plt.plot(dict_d["choclate"],dict_d["Sale"])
plt.show()
```

```python
d["grade"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.show()
```

```python
plt.boxplot([d["weekly_self_study_hours"],d["total_score"]])
plt.show()
```
